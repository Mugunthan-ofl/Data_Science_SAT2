```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from google.colab import files
from IPython.display import display
import zipfile
import os

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully!")
```

```python
uploaded = files.upload()
```

```python
zip_path = next(
    name for name in uploaded
    if name.endswith(".zip")
)

data_dir = Path("/content/air_quality_data")
data_dir.mkdir(exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(data_dir)

print("Dataset extracted successfully!")
print("\nFiles available:")

for file in sorted(data_dir.iterdir()):
    print(file.name)
```

```python
air = pd.read_csv(data_dir / "air_quality.csv")
stations = pd.read_csv(data_dir / "stations.csv")
weather = pd.read_csv(data_dir / "weather.csv")
traffic = pd.read_csv(data_dir / "traffic.csv")
calendar = pd.read_csv(data_dir / "calendar.csv")

print("All datasets loaded successfully!")
```

```python
print("AIR QUALITY DATA")
display(air.head())

print("STATION DATA")
display(stations.head())

print("WEATHER DATA")
display(weather.head())

print("TRAFFIC DATA")
display(traffic.head())

print("CALENDAR DATA")
display(calendar.head())
```

```python
datasets = {
    "Air Quality": air,
    "Stations": stations,
    "Weather": weather,
    "Traffic": traffic,
    "Calendar": calendar
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(name)
    print("=" * 50)

    print("Rows and columns:", df.shape)
    print("Column names:", df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 records:")
    display(df.head())
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    missing = df.isnull().sum()
    percentage = (missing / len(df) * 100).round(2)

    report = pd.DataFrame({
        "Missing_Count": missing,
        "Missing_Percentage": percentage
    })

    print(f"\n{name}")
    display(report)
```

```python
for name, df in datasets.items():
    print(
        f"{name}: "
        f"{df.duplicated().sum()} duplicate rows"
    )
```

```python
air_duplicates = air[
    air.duplicated(keep=False)
].sort_values(["station", "timestamp"])

display(air_duplicates.head(20))
```

```python
print("Original station names:")
print(sorted(air["station"].dropna().unique()))
```

```python
print("Air quality station names:")
print(sorted(air["station"].dropna().unique()))

print("\nStation reference names:")
print(sorted(stations["station"].dropna().unique()))
```

```python
print(air[["pm2_5", "pm10", "no2", "aqi"]].describe())
```

```python
pollutants = ["pm2_5", "pm10", "no2", "aqi"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for col, ax in zip(pollutants, axes.flatten()):
    sns.boxplot(data=air, x=col, ax=ax)
    ax.set_title(f"Distribution of {col.upper()}")

plt.tight_layout()
plt.show()
```

```python
def standardize_columns(df):
    df = df.copy()
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )
    return df

air = standardize_columns(air)
stations = standardize_columns(stations)
weather = standardize_columns(weather)
traffic = standardize_columns(traffic)
calendar = standardize_columns(calendar)

print(air.columns.tolist())
```

```python
for df in [air, weather, traffic]:
    df["timestamp"] = pd.to_datetime(
        df["timestamp"], errors="coerce"
    )

calendar["date"] = pd.to_datetime(
    calendar["date"], errors="coerce"
)

print("Timestamp conversion completed.")
```

```python
for name, df in {
    "Air": air,
    "Weather": weather,
    "Traffic": traffic
}.items():
    print(name, "Invalid timestamps:", df["timestamp"].isna().sum())
```

```python
for df in [air, stations, traffic]:
    df["station"] = (
        df["station"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

print("Standardized station names.")
print(sorted(air["station"].dropna().unique()))
```

```python
print("Duplicates before cleaning:", air.duplicated().sum())

air = air.drop_duplicates().copy()

print("Duplicates after cleaning:", air.duplicated().sum())
print("Rows remaining:", len(air))
```

```python
missing_before = pd.concat({
    "air_quality": air.isna().sum(),
    "weather": weather.isna().sum()
}, axis=1)

display(missing_before)
```

```python
print("Available PM2.5 readings:", air["pm2_5"].count())
print("Missing PM2.5 readings:", air["pm2_5"].isna().sum())
```

```python
air_imputed = air.copy()

for col in ["pm2_5", "pm10", "no2", "aqi"]:
    air_imputed[col] = air_imputed[col].fillna(
        air_imputed[col].median()
    )
```

```python
for col in ["pm2_5", "pm10", "no2", "aqi"]:
    q1 = air[col].quantile(0.25)
    q3 = air[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    air[f"{col}_outlier_flag"] = (
        (air[col] < lower) | (air[col] > upper)
    )

    print(
        col,
        "Potential outliers:",
        int(air[f"{col}_outlier_flag"].sum())
    )
```

```python
merged = air.merge(
    weather,
    on="timestamp",
    how="left",
    validate="many_to_one",
    suffixes=("", "_weather")
)

print("After weather merge:", merged.shape)
```

```python
merged = merged.merge(
    stations,
    on="station",
    how="left",
    validate="many_to_one"
)

print("After station merge:", merged.shape)
print("Missing station locations:", merged["location"].isna().sum())
```

```python
merged = merged.merge(
    traffic,
    on=["timestamp", "station"],
    how="left",
    validate="one_to_one"
)

print("After traffic merge:", merged.shape)
print("Missing traffic records:", merged["traffic_volume"].isna().sum())
```

```python
merged["date"] = merged["timestamp"].dt.normalize()

calendar["date"] = pd.to_datetime(calendar["date"]).dt.normalize()

merged = merged.merge(
    calendar,
    on="date",
    how="left",
    validate="many_to_one"
)

print("Final merged dimensions:", merged.shape)
display(merged.head())
```

```python
print("Rows:", len(merged))
print("Columns:", len(merged.columns))
print("\nMissing values:")
display(merged.isna().sum().sort_values(ascending=False).head(15))

print("\nDuplicate full records:", merged.duplicated().sum())
```

```python
station_summary = (
    merged.groupby("station")
    .agg(
        average_aqi=("aqi", "mean"),
        average_pm25=("pm2_5", "mean"),
        observations=("aqi", "count")
    )
    .sort_values("average_aqi", ascending=False)
)

display(station_summary.round(2))

station_summary["average_aqi"].plot(
    kind="bar", figsize=(10, 5)
)
plt.title("Average AQI by Monitoring Station")
plt.xlabel("Station")
plt.ylabel("Average AQI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
merged["hour"] = merged["timestamp"].dt.hour

hourly_aqi = merged.groupby("hour")["aqi"].mean()

display(hourly_aqi.round(2))

hourly_aqi.plot(figsize=(10, 5), marker="o")
plt.title("Average AQI by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average AQI")
plt.xticks(range(24))
plt.tight_layout()
plt.show()
```

```python
wind_corr = merged[["wind", "pm2_5"]].corr().iloc[0, 1]
print("Wind and PM2.5 correlation:", round(wind_corr, 3))

sns.scatterplot(data=merged, x="wind", y="pm2_5", alpha=0.4)
plt.title("Wind Speed vs PM2.5")
plt.tight_layout()
plt.show()
```

```python
traffic_corr = merged[
    ["traffic_volume", "aqi"]
].corr().iloc[0, 1]

print("Traffic and AQI correlation:", round(traffic_corr, 3))

sns.scatterplot(
    data=merged,
    x="traffic_volume",
    y="aqi",
    alpha=0.4
)
plt.title("Traffic Volume vs AQI")
plt.tight_layout()
plt.show()
```

```python
merged["rain_condition"] = np.where(
    merged["rain"] > 0,
    "Rainy",
    "Dry"
)

rain_summary = merged.groupby("rain_condition").agg(
    average_aqi=("aqi", "mean"),
    average_pm25=("pm2_5", "mean"),
    observations=("aqi", "count")
)

display(rain_summary.round(2))

rain_summary["average_pm25"].plot(kind="bar", figsize=(7, 4))
plt.title("Average PM2.5: Rainy vs Dry")
plt.ylabel("Average PM2.5")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
season_summary = merged.groupby("season").agg(
    average_aqi=("aqi", "mean"),
    average_pm25=("pm2_5", "mean"),
    observations=("aqi", "count")
)

display(season_summary.round(2))

season_summary["average_aqi"].plot(kind="bar", figsize=(7, 4))
plt.title("Average AQI by Season")
plt.ylabel("Average AQI")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
merged["month"] = merged["timestamp"].dt.month
merged["day_of_week"] = merged["timestamp"].dt.day_name()

merged["is_weekend"] = np.where(
    merged["timestamp"].dt.dayofweek >= 5,
    "Weekend",
    "Weekday"
)

merged["traffic_period"] = np.where(
    merged["peak_flag"].fillna(False),
    "Peak",
    "Non-Peak"
)

print("New features created.")
display(
    merged[
        ["timestamp", "hour", "month", "day_of_week",
         "is_weekend", "traffic_period"]
    ].head()
)
```

```python
encoded_data = pd.get_dummies(
    merged,
    columns=["land_use_type"],
    dtype=int
)

display(encoded_data.head())
```
